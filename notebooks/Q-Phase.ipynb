# AirAlert: Q-Phase - Question

AirAlert is a compact machine-learning Streamlit app for urban air-quality risk communication. The project follows the same logic as Noah Roosen's example project "Degrees of No Return": several data layers are transformed into a clear, interactive risk signal for non-expert users. Instead of climate flooding and heat risk, AirAlert focuses on air pollution in European cities.

The guiding idea is simple: raw pollution and weather values are difficult to interpret for many users. AirAlert translates them into a directly understandable risk class: Low, Medium or High.

## Central research question

How can air-quality indicators, weather conditions and city-context variables be combined in a machine-learning Streamlit app to classify urban air-quality risk in a way that is understandable for non-expert users?

The project is therefore not only a prediction task. It is also a knowledge-transfer task: the app should explain why the model classifies a scenario as low, medium or high risk.

## Thematic research questions

- Which pollutant indicators, such as PM2.5, PM10, NO2 and O3, contribute most strongly to the predicted air-quality risk?
- How do weather factors such as wind speed, temperature and humidity influence the risk class?
- How can city-context variables such as traffic intensity, green space and industrial activity make the prediction more realistic?
- How can the result be visualized in Streamlit so that users can compare cities, adjust scenarios and understand the model output?

## Target group and use case

The app is designed for students, city residents, commuters and local decision makers who need an understandable overview instead of raw technical tables. A commuter might ask whether outdoor activity is risky in a specific month. A student might use the app to understand how weather and pollutants interact. A local decision maker might use it as an educational demonstration of data-driven environmental risk communication.

The app is not a medical or regulatory warning system. It is a classroom ML prototype that demonstrates a complete data-science workflow.

## Data foundation

The app is based on three data layers:

1. Air-quality layer: PM2.5, PM10, NO2, O3 and an AQI-like target class.
2. Weather layer: temperature, humidity and wind speed.
3. City-context layer: population, traffic, green-space and industry indicators.

For the first stable version, deterministic demo data is generated inside the project. This makes the app reliable in PyCharm, Google Colab and Streamlit Cloud without API keys or fragile downloads. In the final explanation, these layers are mapped to real public data sources such as OpenAQ, Open-Meteo and city metadata datasets.

## Success criteria

The project is successful if:

- the Streamlit app runs without errors,
- the user can select a city and month,
- the user can adjust scenario inputs,
- the app returns a clear Low/Medium/High risk class,
- the model evaluation is visible,
- the data-source logic is transparent,
- the QUA3CK process is documented from question to knowledge transfer.

## Relation to Noah's example project

Noah's app translates global climate data into local climate-risk views. AirAlert translates pollution, weather and city-context data into urban air-quality risk. Both projects follow the same structure:

multiple datasets -> ML model -> interactive Streamlit dashboard -> understandable risk output.

AirAlert is intentionally smaller so that the logic can be explained clearly and the app remains stable for a beginner-level course submission.
